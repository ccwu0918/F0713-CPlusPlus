### 幻燈片 5：變數與資料型態

# 📦 變數 (Variable) 與資料型態 (Data Type)

| 圖示 | 型態 | 大小 (一般) | 範例 | 電機情境 |
|------|------|------|------|------|
| 🔢 | `int` | 4 bytes | `int resistance = 330;` | 電阻值 (Ω) |
| 📊 | `double` | 8 bytes | `double voltage = 5.0;` | 電壓 (V)，精度較高 |
| 📉 | `float` | 4 bytes | `float current = 0.015f;` | 電流 (A)，精度較低 |
| 🔤 | `char` | 1 byte | `char type = 'R';` | 元件代號 R/L/C |
| 💡 | `bool` | 1 byte | `bool ledOn = true;` | 開關狀態 ON/OFF |
| 📝 | `string` | 不固定 | `string part = "1N4148";` | 元件型號 |

#### 💡 電機系比喻
變數就像麵包板 (Breadboard) 上的插孔，資料型態決定了「這個孔能插什麼元件」。
`sizeof()` 可以查詢每種型態佔多少記憶體 (bytes)。

````python
%%cuda
#include <iostream>
#include <string>
using namespace std;

int main() {
    int    resistance = 330;      // 電阻值 330Ω
    double voltage    = 5.0;      // 電源電壓 5V
    float  current    = 0.015f;   // 電流 15mA
    char   type       = 'R';      // 元件代號 R = 電阻
    bool   ledOn      = true;     // LED 狀態
    string part       = "1N4148"; // 二極體型號

    cout << "電阻值：" << resistance << " Ω" << endl;
    cout << "電壓：" << voltage << " V" << endl;
    cout << "電流：" << current << " A" << endl;
    cout << "元件代號：" << type << endl;
    cout << "LED 狀態：" << ledOn << " (1=ON, 0=OFF)" << endl;
    cout << "二極體型號：" << part << endl;

    cout << "--- sizeof (bytes) ---" << endl;
    cout << "int: " << sizeof(int) << ", double: " << sizeof(double)
         << ", float: " << sizeof(float) << ", char: " << sizeof(char)
         << ", bool: " << sizeof(bool) << endl;
    return 0;
}
````

### 幻燈片 6：cout 格式化輸出

# 📤 cout 與 `<iomanip>`

| 語法 | 說明 | 類比 C 語言 |
|------|------|------|
| `fixed` | 固定小數點表示 | `%f` |
| `setprecision(n)` | 搭配 fixed：小數點後 n 位 | `%.nf` |
| `setw(n)` | 欄位寬度 n | `%nd` |
| `endl` / `"\n"` | 換行 | `\n` |

#### 💡 電機系比喻
就像三用電表 (Multimeter) 的量程切換：同一個數值，可以選擇顯示到小數點後幾位。

````python
%%cuda
#include <iostream>
#include <iomanip>   // setprecision, setw, fixed
using namespace std;

int main() {
    double v = 3.3;
    double r = 1000.0;
    double i = v / r;          // 歐姆定律 I = V / R

    cout << "預設顯示：" << i << " A" << endl;
    cout << fixed << setprecision(4);
    cout << "小數 4 位：" << i << " A" << endl;
    cout << "換算毫安：" << i * 1000 << " mA" << endl;

    // 用 setw 對齊，做出簡易量測表
    cout << setprecision(2);
    cout << setw(8) << "V(V)" << setw(10) << "I(mA)" << endl;
    for (double volt = 1.0; volt <= 5.0; volt += 2.0) {
        cout << setw(8) << volt << setw(10) << volt / r * 1000 << endl;
    }
    return 0;
}
````

### 幻燈片 7：cin 輸入

# 📥 cin 從鍵盤讀取資料

- `cin >> x;` 的 `>>` 稱為擷取運算子 (Extraction Operator)
- 不需要 `&`！（比 C 的 `scanf` 安全很多）
- 可以連續輸入：`cin >> v >> r;`（空白或換行分隔）

#### ⚡ 電機系範例：歐姆定律計算機
輸入電壓 V 與電阻 R，計算電流 I 與功率 P。

> 因為 `%%cuda` 無法互動輸入，這裡改用 **writefile → g++ 編譯 → 執行** 三步驟。

````python
%%writefile ohm.cpp
#include <iostream>
#include <iomanip>
using namespace std;

int main() {
    double voltage, resistance;

    cout << "請輸入電壓 (V) 與電阻 (Ω)：";
    cin >> voltage >> resistance;

    double current = voltage / resistance;   // I = V / R
    double power   = voltage * current;      // P = V * I

    cout << fixed << setprecision(3) << endl;
    cout << "電流 I = " << current * 1000 << " mA" << endl;
    cout << "功率 P = " << power * 1000 << " mW" << endl;

    if (power > 0.25) {
        cout << "⚠️ 超過 1/4W 電阻的額定功率 (Power Rating)！" << endl;
    }
    return 0;
}
````

### 幻燈片 8：算術運算子

# ➕ 算術運算子 (Arithmetic Operators)

| 運算子 | 說明 | 範例 |
|------|------|------|
| `+ - * /` | 加減乘除 | `v * i` |
| `%` | 取餘數 (Modulus)，只能用於整數 | `ms % 1000` |
| `++` / `--` | 遞增 / 遞減 | `count++` |
| `+= -= *= /=` | 複合指定 (Compound Assignment) | `total += i` |

#### ⚠️ 整數除法陷阱 (Integer Division)
`int / int` 的結果還是 `int`，小數會被**直接捨去**！
分壓電路 (Voltage Divider) 計算時，一不小心就會算出 0 V。

````python
%%cuda
#include <iostream>
using namespace std;

int main() {
    // 分壓電路：Vout = Vin * R2 / (R1 + R2)
    int vin = 5, r1 = 10000, r2 = 4700;

    int    wrong = vin * (r2 / (r1 + r2));            // 4700/14700 → 0 ！
    double right = vin * (double)r2 / (r1 + r2);      // 先轉成 double

    cout << "錯誤寫法 Vout = " << wrong << " V" << endl;
    cout << "正確寫法 Vout = " << right << " V" << endl;

    // % 取餘數：把毫秒換算成 秒 + 毫秒
    int ms = 3725;
    cout << ms << " ms = " << ms / 1000 << " s + " << ms % 1000 << " ms" << endl;

    // ++ 與 +=：計數器
    int pulses = 0;
    pulses++;
    pulses += 10;
    cout << "脈衝計數 (Pulse Count) = " << pulses << endl;
    return 0;
}
````

**輸出：**
```
錯誤寫法 Vout = 0 V
正確寫法 Vout = 1.59864 V
3725 ms = 3 s + 725 ms
脈衝計數 (Pulse Count) = 11
```

### 幻燈片 9：比較、邏輯運算子與 bool

# ⚖️ 比較 (Comparison) 與邏輯 (Logical) 運算子

| 比較 | 意義 | 邏輯 | 意義 | 數位電路對應 |
|------|------|------|------|------|
| `==` | 等於 | `&&` | 且 (AND) | AND Gate |
| `!=` | 不等於 | `\|\|` | 或 (OR) | OR Gate |
| `> <` | 大於、小於 | `!` | 非 (NOT) | NOT Gate (Inverter) |
| `>= <=` | 大於等於、小於等於 | | | |

結果是 `bool`：`true`(1) 或 `false`(0)。加上 `boolalpha` 可以直接印出 true/false。

#### 💡 電機系比喻
這就是你在「數位邏輯 (Digital Logic)」學的邏輯閘！下面用程式印出 AND / OR / XOR 真值表 (Truth Table)。

````python
%%cuda
#include <iostream>
using namespace std;

int main() {
    cout << " A B | AND OR XOR" << endl;
    cout << "-----+-----------" << endl;
    for (int a = 0; a <= 1; a++) {
        for (int b = 0; b <= 1; b++) {
            bool A = a, B = b;
            cout << " " << A << " " << B << " |  "
                 << (A && B) << "   " << (A || B) << "   " << (A != B) << endl;
        }
    }

    // 實際應用：溫度與電流同時正常，系統才算安全
    double temp = 65.0, amp = 2.5;
    bool safe = (temp < 80.0) && (amp < 3.0);
    cout << boolalpha << "\n系統安全？ " << safe << endl;
    return 0;
}
````

### 幻燈片 10：位元運算 — 電機系必修！

# 🔌 位元運算子 (Bitwise Operators)

微控制器 (MCU) 用一個 8-bit **暫存器 (Register)** 控制 8 支 GPIO 腳位，每個位元 (bit) 代表一支腳。

| 運算子 | 名稱 | 用途 | 寫法 |
|------|------|------|------|
| `\|` | Bitwise OR | 把某一位設為 1 (Set) | `reg \|= (1 << n);` |
| `&` + `~` | AND + NOT | 把某一位清為 0 (Clear) | `reg &= ~(1 << n);` |
| `^` | XOR | 反轉某一位 (Toggle) | `reg ^= (1 << n);` |
| `&` | AND | 讀取某一位 (Read) | `(reg >> n) & 1` |
| `<<` `>>` | 左移 / 右移 (Shift) | 產生遮罩 (Mask) | `1 << 3` = `00001000` |

`bitset<8>` 可以把數字直接印成二進位，方便觀察。

````python
%%cuda
#include <iostream>
#include <bitset>
using namespace std;

int main() {
    unsigned char portB = 0b00000000;   // 8-bit GPIO 暫存器，全部 LOW

    portB |= (1 << 0);                  // Set：點亮 LED0
    portB |= (1 << 3);                  // Set：點亮 LED3
    cout << "Set  LED0, LED3 : " << bitset<8>(portB) << endl;

    portB ^= (1 << 7);                  // Toggle：反轉 LED7
    cout << "Toggle LED7     : " << bitset<8>(portB) << endl;

    portB &= ~(1 << 0);                 // Clear：熄滅 LED0
    cout << "Clear LED0      : " << bitset<8>(portB) << endl;

    int pin3 = (portB >> 3) & 1;        // Read：讀取第 3 腳
    cout << "Pin3 狀態 = " << pin3 << endl;

    cout << "十進位值 = " << (int)portB << ", 十六進位 = 0x" << hex << (int)portB << endl;
    return 0;
}
````

**輸出：**
```
Set  LED0, LED3 : 00001001
Toggle LED7     : 10001001
Clear LED0      : 10001000
Pin3 狀態 = 1
十進位值 = 136, 十六進位 = 0x88
```

### 幻燈片 11：cmath 數學函式

# 📐 `<cmath>` 與 max / min

| 函式 | 說明 | 電機情境 |
|------|------|------|
| `sqrt(x)` | 平方根 | 有效值 RMS = Vp / √2 |
| `pow(x, y)` | x 的 y 次方 | 功率 P = I² R |
| `abs(x)` / `fabs(x)` | 絕對值 | 量測誤差 |
| `exp(x)` | eˣ | RC 充放電 |
| `log10(x)` | 以 10 為底對數 | 分貝 dB = 20 log₁₀(Vout/Vin) |
| `max(a,b)` / `min(a,b)` | 最大 / 最小值 | 限流、限壓 |

````python
%%cuda
#include <iostream>
#include <cmath>
#include <algorithm>   // max, min
using namespace std;

int main() {
    const double PI = 3.14159265358979;

    // 台灣市電：有效值 110V，峰值 Vp = Vrms * √2
    double vrms = 110.0;
    cout << "市電峰值 Vp = " << vrms * sqrt(2.0) << " V" << endl;

    // 功率 P = I^2 * R
    double I = 0.5, R = 20.0;
    cout << "功率 P = " << pow(I, 2) * R << " W" << endl;

    // 放大器增益 (Gain) 換算 dB
    double vin = 0.1, vout = 2.0;
    cout << "電壓增益 = " << 20 * log10(vout / vin) << " dB" << endl;

    // 角頻率 ω = 2πf
    double f = 60.0;
    cout << "60Hz 角頻率 ω = " << 2 * PI * f << " rad/s" << endl;

    // 量測誤差
    double measured = 4.87, expected = 5.0;
    cout << "誤差 = " << fabs(measured - expected) / expected * 100 << " %" << endl;

    // 限流保護：電流不超過 2A
    double request = 3.2;
    cout << "實際輸出電流 = " << min(request, 2.0) << " A" << endl;
    return 0;
}
````

### 幻燈片 15：switch-case 多重選擇

# 🎨 switch-case

- 適合處理「固定值」的多重選擇（`int`、`char`，不能用 `double` 或 `string`）
- 每個 `case` 結尾記得加 `break`，否則會「穿透 (Fall-through)」到下一個 case
- `default` 處理沒有對應的情況

#### ⚡ 電機系範例：電阻色碼 (Resistor Color Code)
四環電阻：第 1、2 環是數字，第 3 環是倍率 (Multiplier)。例如 **黃 紫 棕** = 47 × 10¹ = 470Ω。

````python
%%cuda
#include <iostream>
using namespace std;

// 色碼字元 → 數字
int colorDigit(char c) {
    switch (c) {
        case 'K': return 0;   // Black  黑
        case 'N': return 1;   // browN  棕
        case 'R': return 2;   // Red    紅
        case 'O': return 3;   // Orange 橙
        case 'Y': return 4;   // Yellow 黃
        case 'G': return 5;   // Green  綠
        case 'B': return 6;   // Blue   藍
        case 'V': return 7;   // Violet 紫
        case 'A': return 8;   // grAy   灰
        case 'W': return 9;   // White  白
        default:  return -1;  // 無效色碼
    }
}

void decode(char b1, char b2, char b3) {
    int d1 = colorDigit(b1), d2 = colorDigit(b2), m = colorDigit(b3);
    if (d1 < 0 || d2 < 0 || m < 0) {
        cout << "色碼錯誤！" << endl;
        return;
    }
    long ohm = d1 * 10 + d2;
    for (int i = 0; i < m; i++) ohm *= 10;
    cout << b1 << b2 << b3 << " = " << ohm << " Ω" << endl;
}

int main() {
    decode('Y', 'V', 'N');   // 黃紫棕 = 470 Ω
    decode('N', 'K', 'O');   // 棕黑橙 = 10000 Ω
    decode('R', 'R', 'R');   // 紅紅紅 = 2200 Ω
    decode('X', 'K', 'O');   // 無效
    return 0;
}
````

#### ⚡ 電機系範例：RC 電路充電
電容電壓 $V_C(t) = V_S\,(1 - e^{-t/RC})$。
用 while 迴圈每 0.5 ms 計算一次，直到充到電源電壓的 99%，看看需要幾個時間常數 τ = RC。

````python
%%cuda
#include <iostream>
#include <cmath>
#include <iomanip>
using namespace std;

int main() {
    double Vs  = 5.0;        // 電源電壓 5V
    double R   = 1000.0;     // 1kΩ
    double C   = 1e-6;       // 1μF
    double tau = R * C;      // 時間常數 τ = 1 ms
    double t   = 0.0, dt = 0.0005;
    double Vc  = 0.0;

    cout << fixed << setprecision(3);
    while (Vc < 0.99 * Vs) {
        Vc = Vs * (1 - exp(-t / tau));
        cout << "t = " << t * 1000 << " ms, Vc = " << Vc << " V" << endl;
        t += dt;
    }
    cout << "約經過 " << (t - dt) / tau << " τ 充到 99%" << endl;
    return 0;
}
````

#### ⚡ 電機系範例：開機自我檢測 (POST, Power-On Self-Test)
裝置開機一定要先檢測一次；檢測失敗就重試，最多 3 次。

````python
%%cuda
#include <iostream>
using namespace std;

int main() {
    // 模擬三次量測到的電源電壓（第 3 次才穩定）
    double readings[] = {4.20, 4.55, 4.98};
    int attempt = 0;
    bool ok;

    do {
        double v = readings[attempt];
        ok = (v >= 4.75 && v <= 5.25);     // 5V ± 5% 才算合格
        cout << "第 " << attempt + 1 << " 次檢測：" << v << " V → "
             << (ok ? "PASS" : "FAIL") << endl;
        attempt++;
    } while (!ok && attempt < 3);

    cout << (ok ? "✅ 開機成功" : "❌ 電源異常，停止開機") << endl;
    return 0;
}
````

### 幻燈片 19：break 與 continue

# ⏹️ break 跳出迴圈 / ⏭️ continue 略過本次

| 關鍵字 | 作用 | 電機情境 |
|------|------|------|
| `break` | 立刻**離開**整個迴圈 | 溫度過高 → 緊急停機 |
| `continue` | **略過**這一次，直接進下一輪 | 感測器斷線 (讀值 -1) → 略過該通道 |

````python
%%cuda
#include <iostream>
using namespace std;

int main() {
    // 8 個溫度感測通道的讀值 (°C)，-1 代表感測器斷線
    double temp[8] = {42.5, 45.0, -1, 51.2, 88.6, 47.3, 49.9, 40.1};
    double sum = 0;
    int count = 0;

    for (int ch = 0; ch < 8; ch++) {
        if (temp[ch] < 0) {
            cout << "CH" << ch << "：感測器斷線，略過" << endl;
            continue;                               // 略過這個通道
        }
        if (temp[ch] > 85.0) {
            cout << "CH" << ch << "：" << temp[ch] << "°C 過熱！緊急停機" << endl;
            break;                                  // 直接停止掃描
        }
        cout << "CH" << ch << "：" << temp[ch] << "°C" << endl;
        sum += temp[ch];
        count++;
    }
    cout << "正常通道平均溫度 = " << sum / count << "°C" << endl;
    return 0;
}
````

````python
%%cuda
#include <iostream>
using namespace std;

int main() {
    // 愛心圖案，每一列一個 byte
    unsigned char heart[8] = {
        0b00000000,
        0b01100110,
        0b11111111,
        0b11111111,
        0b11111111,
        0b01111110,
        0b00111100,
        0b00011000
    };

    for (int row = 0; row < 8; row++) {          // 外層：掃描每一列
        for (int col = 7; col >= 0; col--) {     // 內層：檢查每一個位元
            if ((heart[row] >> col) & 1)
                cout << "● ";
            else
                cout << "· ";
        }
        cout << endl;
    }
    return 0;
}
````

### 幻燈片 21：階段二測驗

# 🔍 找錯誤：永遠成立的 if

#### 這段程式明明電壓是 3.3V，為什麼還會顯示「斷電」？

```cpp
int voltage = 3;
if (voltage = 0) {
    cout << "斷電！" << endl;
} else {
    cout << "電源正常" << endl;
}
```

執行看看，並觀察 `voltage` 最後的值。

````python
%%cuda
#include <iostream>
using namespace std;

int main() {
    int voltage = 3;
    if (voltage = 0) {
        cout << "斷電！" << endl;
    } else {
        cout << "電源正常" << endl;
    }
    cout << "voltage 現在是 " << voltage << endl;
    return 0;
}
````

---

#### ✅ 解答

**問題：** `=` 是「指定 (Assignment)」，`==` 才是「比較 (Comparison)」！
`if (voltage = 0)` 會先把 voltage **改成 0**，再拿 0 當條件 → 0 視為 false，所以進入 else。
這個 bug 很陰險：程式不會報錯，但**偷偷改掉了你的變數**。

**修正：** `if (voltage == 0)`

**防呆技巧 (Yoda Condition)：** 寫成 `if (0 == voltage)`，萬一少打一個 `=` 變成 `0 = voltage`，編譯器會直接報錯。

#### 💡 就像量測時把電表切到「電流檔」去量電壓 — 不但讀不到正確值，還可能把電路短路！

#### ⚡ 電機系範例：解析 BOM 表 (Bill of Materials) 的元件標記
格式為 `元件代號-數值`，例如 `R1-470`、`C3-100`。

````python
%%cuda
#include <iostream>
#include <string>
using namespace std;

int main() {
    string bom[4] = {"R1-470", "R2-10000", "C3-100", "LED1-RED"};

    for (int i = 0; i < 4; i++) {
        string item = bom[i];
        size_t dash = item.find("-");             // 找到 '-' 的位置
        string ref  = item.substr(0, dash);        // 元件代號
        string val  = item.substr(dash + 1);       // 數值部分

        cout << "標記 " << item << "（長度 " << item.length() << "）→ ";
        if (ref[0] == 'R') {
            int ohm = stoi(val);
            cout << "電阻 " << ref << " = " << ohm << " Ω";
            if (ohm >= 1000) cout << " = " << ohm / 1000.0 << " kΩ";
        } else if (ref[0] == 'C') {
            cout << "電容 " << ref << " = " << val << " nF";
        } else {
            cout << "其他元件 " << ref << "（" << val << "）";
        }
        cout << endl;
    }

    string label = "Vout=" + to_string(3.3) + "V";   // to_string 預設保留 6 位小數
    cout << label << endl;
    return 0;
}
````

### 幻燈片 25：vector 動態陣列

# 📈 vector — 會自己長大的陣列

| 操作 | 語法 |
|------|------|
| 宣告 | `vector<double> samples;` |
| 加到尾端 | `samples.push_back(3.3);` |

#### 💡 電機系比喻
陣列像是固定 8 個通道的類比多工器；vector 像示波器的記憶體深度 (Memory Depth)，取樣多少就存多少。

#### ⚡ 電機系範例：計算取樣訊號的平均值、峰對峰值 (Vpp)、有效值 (RMS)

````python
%%cuda
#include <iostream>
#include <vector>
#include <cmath>
using namespace std;

int main() {
    const double PI = 3.14159265358979;
    vector<double> samples;

    // 模擬 ADC 取樣一個週期的正弦波：振幅 2V、40 個取樣點
    for (int n = 0; n < 40; n++) {
        samples.push_back(2.0 * sin(2 * PI * n / 40));
    }

    double sum = 0, sumSq = 0;
    double vmax = samples[0], vmin = samples[0];
    for (double v : samples) {            // 範圍 for 迴圈
        sum   += v;
        sumSq += v * v;
        if (v > vmax) vmax = v;
        if (v < vmin) vmin = v;
    }

    cout << "取樣點數 = " << samples.size() << endl;
    cout << "平均值 (DC) ≈ " << round(sum / samples.size() * 1000) / 1000 << " V" << endl;
    cout << "峰對峰值 Vpp = " << vmax - vmin << " V" << endl;
    cout << "有效值 Vrms = " << sqrt(sumSq / samples.size()) << " V" << endl;
    cout << "理論值 Vp/√2 = " << 2.0 / sqrt(2.0) << " V" << endl;
    return 0;
}
````

#### ⚡ 電機系範例：導線功率損耗 P = I²R

### 幻燈片 28：函式多載

# 🔀 函式多載 (Function Overloading)

**同一個函式名稱**，只要參數的**個數或型態不同**，就可以定義好幾個版本，編譯器會自動挑選正確的那一個。
（C 語言不能這樣做！）

#### ⚡ 電機系範例：並聯電阻 `parallel()`
- 2 顆電阻：$R = \dfrac{R_1R_2}{R_1+R_2}$
- 3 顆電阻、任意多顆電阻（vector）：$\dfrac{1}{R} = \sum \dfrac{1}{R_i}$

````python
%%cuda
#include <iostream>
#include <vector>
using namespace std;

double parallel(double r1, double r2) {
    return r1 * r2 / (r1 + r2);
}

double parallel(double r1, double r2, double r3) {
    return 1.0 / (1.0 / r1 + 1.0 / r2 + 1.0 / r3);
}

double parallel(const vector<double>& rs) {
    double g = 0;                  // 電導 (Conductance) G = 1/R，單位 S
    for (double r : rs) g += 1.0 / r;
    return 1.0 / g;
}

int main() {
    cout << "2 顆 1kΩ 並聯 = " << parallel(1000, 1000) << " Ω" << endl;
    cout << "3 顆 1kΩ 並聯 = " << parallel(1000, 1000, 1000) << " Ω" << endl;

    vector<double> bank = {1000, 2200, 4700, 10000};
    cout << "4 顆電阻並聯 = " << parallel(bank) << " Ω" << endl;
    return 0;
}
````

### 幻燈片 29：階段三測驗

#### ✅ 解答

**問題：** C++ 編譯器**由上往下**讀程式。讀到 `main()` 裡的 `power(...)` 時，它還不知道 `power` 是什麼 → `'power' was not declared in this scope`。

**修正方法（擇一）：**
1. 把 `power()` 整個函式搬到 `main()` 上面
2. 在 `main()` 上面加函式原型：`double power(double v, double i);`

#### 💡 就像畫電路圖時用了一顆 IC，卻沒在零件表 (BOM) 裡登記，採購的人根本不知道要買什麼！

### 幻燈片 31：參考 Reference

# 🏷️ 參考：變數的別名

```cpp
double vcc = 5.0;
double &supply = vcc;   // supply 就是 vcc 的另一個名字
```

- 宣告時**一定要初始化**，之後不能改綁別的變數
- 改 `supply` 就等於改 `vcc`，兩者是**同一塊記憶體**

#### 💡 電機系比喻
電路圖上的網路標籤 (Net Label)：在不同頁面標了兩個 `VCC`，它們其實是**同一條線**，任何一端電壓改變，另一端也跟著變。

````python
%%cuda
#include <iostream>
using namespace std;

int main() {
    double vcc = 5.0;
    double &supply = vcc;          // 建立參考

    cout << "vcc = " << vcc << ", supply = " << supply << endl;

    supply = 3.3;                  // 透過參考修改
    cout << "改 supply 後 → vcc = " << vcc << endl;

    cout << "vcc 的位址    = " << &vcc << endl;
    cout << "supply 的位址 = " << &supply << "（相同！）" << endl;
    return 0;
}
````

### 幻燈片 32：傳值 vs 傳參考 vs 傳指標

# 📮 三種參數傳遞方式

| 方式 | 寫法 | 函式能否改到原變數？ | 比喻 |
|------|------|------|------|
| 傳值 (Pass by Value) | `void f(double x)` | ❌ 不能（拿到複本） | 把量測報告「影印」一份給你 |
| 傳參考 (Pass by Reference) | `void f(double &x)` | ✅ 可以 | 直接把探棒接到原電路上 |
| 傳指標 (Pass by Pointer) | `void f(double *x)` | ✅ 可以（呼叫時要寫 `&x`） | 給你電路板的「座標」自己去找 |

#### ⚡ 電機系範例：感測器校正 (Calibration)
溫度感測器讀值有 +1.5°C 的偏移 (Offset)，寫一個校正函式把讀值修正回來。

````python
%%cuda
#include <iostream>
using namespace std;

void calibrateByValue(double t)      { t = t - 1.5; }
void calibrateByRef(double &t)       { t = t - 1.5; }
void calibrateByPtr(double *t)       { *t = *t - 1.5; }

// 用參考一次「回傳」兩個結果：最大值與最小值
void minMax(const double data[], int n, double &mn, double &mx) {
    mn = mx = data[0];
    for (int i = 1; i < n; i++) {
        if (data[i] < mn) mn = data[i];
        if (data[i] > mx) mx = data[i];
    }
}

int main() {
    double a = 26.5, b = 26.5, c = 26.5;

    calibrateByValue(a);
    calibrateByRef(b);
    calibrateByPtr(&c);

    cout << "傳值   後：" << a << "°C（沒變！）" << endl;
    cout << "傳參考 後：" << b << "°C" << endl;
    cout << "傳指標 後：" << c << "°C" << endl;

    double temps[5] = {24.1, 25.3, 23.8, 26.0, 24.7};
    double lo, hi;
    minMax(temps, 5, lo, hi);
    cout << "最低 " << lo << "°C，最高 " << hi << "°C" << endl;
    return 0;
}
````

### 幻燈片 33：指標基礎

# 📬 指標 (Pointer)：存放「位址」的變數

| 符號 | 位置 | 意義 |
|------|------|------|
| `int *p;` | 宣告時 | p 是一個指向 int 的指標 |
| `&x` | 運算式中 | 取址 (Address-of)：x 的記憶體位址 |
| `*p` | 運算式中 | 取值 / 解參考 (Dereference)：p 指到的內容 |
| `nullptr` | | 空指標，代表「沒有指向任何東西」 |

#### 💡 電機系比喻：記憶體映射 I/O (Memory-Mapped I/O)
在微控制器裡，每個周邊暫存器都有固定的**位址**。韌體工程師就是透過「指向該位址的指標」來控制硬體，例如 `*GPIO_OUT = 0xFF;`。

````python
%%cuda
#include <iostream>
using namespace std;

int main() {
    int adcValue = 512;          // 10-bit ADC 讀值
    int *p = &adcValue;          // p 存放 adcValue 的位址

    cout << "adcValue 的值   = " << adcValue << endl;
    cout << "adcValue 的位址 = " << &adcValue << endl;
    cout << "p 存的位址      = " << p << endl;
    cout << "*p 取出的值     = " << *p << endl;

    *p = 1023;                   // 透過指標寫入（像寫入暫存器）
    cout << "透過 *p 寫入後，adcValue = " << adcValue << endl;

    // 換算成電壓：Vref = 5V，10-bit 解析度 → 0~1023
    double volt = *p * 5.0 / 1023;
    cout << "對應電壓 = " << volt << " V" << endl;

    int *q = nullptr;            // 好習慣：不知道指向哪裡就設成 nullptr
    if (q == nullptr) cout << "q 尚未指向任何位址" << endl;
    return 0;
}
````

### 幻燈片 34：指標與陣列

# 📊 指標與陣列

陣列名稱就是**第一個元素的位址**：

```
data[i]  等同於  *(data + i)
```

指標 `+1` 不是加 1 個 byte，而是**往後移一個元素**（`int` 移 4 bytes、`double` 移 8 bytes）。

#### 💡 電機系比喻
就像 DMA (Direct Memory Access) 控制器的位址暫存器：每搬完一筆資料，位址自動往後遞增一個資料寬度。

````python
%%cuda
#include <iostream>
using namespace std;

int main() {
    int data[5] = {100, 200, 300, 400, 500};
    int *p = data;                     // 等同 &data[0]

    cout << "data[2]     = " << data[2] << endl;
    cout << "*(data + 2) = " << *(data + 2) << endl;
    cout << "*(p + 2)    = " << *(p + 2) << endl;

    cout << "\n用指標走訪 (Traverse) 陣列：" << endl;
    for (int *ptr = data; ptr < data + 5; ptr++) {
        cout << "位址 " << ptr << " → " << *ptr << endl;
    }
    cout << "（每次位址增加 " << sizeof(int) << " bytes）" << endl;
    return 0;
}
````

### 幻燈片 35：new 與 delete

# 💾 動態記憶體配置

| C 語言 | C++ |
|------|------|
| `int *p = (int*)malloc(sizeof(int));` | `int *p = new int;` |
| `free(p);` | `delete p;` |
| `int *a = (int*)malloc(n * sizeof(int));` | `int *a = new int[n];` |
| `free(a);` | `delete[] a;` ← 陣列要加 `[]`！ |

#### ⚠️ 重要原則
- 每一個 `new` 都要對應一個 `delete`，否則**記憶體洩漏 (Memory Leak)**
- `delete` 後把指標設為 `nullptr`，避免**懸空指標 (Dangling Pointer)**
- 實務上大多用 `vector` 取代 `new[]`，它會自動管理記憶體

#### ⚡ 電機系範例：執行時才決定取樣數的資料緩衝區 (Buffer)

````python
%%cuda
#include <iostream>
using namespace std;

int main() {
    int n = 6;                              // 執行時才決定取樣點數
    double *buffer = new double[n];         // 動態配置
    cout << "配置 " << n << " 個 double 的緩衝區" << endl;

    // 模擬電池放電：每次量測下降 0.1V
    for (int i = 0; i < n; i++) {
        buffer[i] = 4.2 - 0.1 * i;
    }

    for (int i = 0; i < n; i++) {
        cout << "第 " << i << " 筆：" << buffer[i] << " V" << endl;
    }

    delete[] buffer;                        // 陣列用 delete[]
    buffer = nullptr;                       // 避免懸空指標
    cout << "緩衝區已釋放" << endl;
    return 0;
}
````

### 幻燈片 36：階段四測驗

# 🔍 找錯誤：指標的三大地雷

```cpp
int *a = new int[5];
delete a;              // ①

int *p = new int(10);
delete p;
cout << *p;            // ②

int *q;
*q = 5;                // ③
```

這三行分別犯了什麼錯？（這類錯誤**通常能編譯成功**，但執行時可能當機或出現亂碼，屬於「未定義行為 (Undefined Behavior)」，所以這裡只展示正確版本。）

---

#### ✅ 解答

| 編號 | 錯誤 | 正確寫法 |
|------|------|------|
| ① | 用 `new[]` 配置卻用 `delete` 釋放 | `delete[] a;` |
| ② | `delete` 後還使用 → 懸空指標 (Dangling Pointer) | `delete p; p = nullptr;`，之後不再使用 |
| ③ | 指標沒初始化就寫入 → 野指標 (Wild Pointer) | `int x; int *q = &x;` 或 `int *q = nullptr;` |

#### 💡 就像拿探棒去戳一個「不知道接到哪裡」的測試點：可能沒事，也可能直接短路燒板！

````python
%%cuda
#include <iostream>
using namespace std;

int main() {
    int *a = new int[5];
    delete[] a;                 // ① 修正：陣列用 delete[]
    a = nullptr;

    int *p = new int(10);
    cout << "*p = " << *p << endl;
    delete p;                   // ② 修正：用完才釋放，並設 nullptr
    p = nullptr;

    int x = 0;
    int *q = &x;                // ③ 修正：指標先指向合法的變數
    *q = 5;
    cout << "x = " << x << endl;
    return 0;
}
````

# 🧩 物件導向程式設計 (OOP, Object-Oriented Programming)

## 把「資料」和「操作資料的函式」包在一起

### 幻燈片 38：類別與物件

# 🏗️ Class 與 Object

- **類別 (Class)** = 藍圖 / 規格書 (Datasheet)：定義「有哪些屬性、能做什麼」
- **物件 (Object)** = 依照藍圖做出來的實體 (Instance)：一顆真正的電阻
- 屬性 (Attribute) = 成員變數；方法 (Method) = 成員函式
- 用點運算子 `.` 存取成員：`r1.ohm`、`r1.current(5.0)`
- class 定義結尾要加**分號** `;`

```
   Class Resistor (規格書)            Objects (實體零件)
  ┌────────────────────┐        ┌──────────┐ ┌──────────┐
  │ ohm, tolerance,    │  ───▶  │ r1: 220Ω │ │ r2: 10kΩ │
  │ watt               │        │  5%,1/4W │ │  1%,1/8W │
  │ current(), power() │        └──────────┘ └──────────┘
  └────────────────────┘
```

````python
%%cuda
#include <iostream>
using namespace std;

class Resistor {                 // 類別：電阻的規格書
  public:                        // 存取修飾子：外部可以存取
    double ohm;                  // 阻值 (Ω)
    double tolerance;            // 誤差 (%)
    double watt;                 // 額定功率 (W)

    // 成員函式（定義在類別內）
    double current(double volt) {
        return volt / ohm;
    }
    double power(double volt) {
        return volt * volt / ohm;
    }
    void info() {
        cout << ohm << "Ω ±" << tolerance << "% " << watt << "W";
    }
};                               // ← 別忘了分號！

int main() {
    Resistor r1;                 // 物件 1
    r1.ohm = 220; r1.tolerance = 5; r1.watt = 0.25;

    Resistor r2;                 // 物件 2
    r2.ohm = 10000; r2.tolerance = 1; r2.watt = 0.125;

    double v = 5.0;
    r1.info();
    cout << " → I = " << r1.current(v) * 1000 << " mA, P = " << r1.power(v) << " W";
    cout << (r1.power(v) > r1.watt ? "  ⚠️ 超過額定功率！" : "  ✅") << endl;

    r2.info();
    cout << " → I = " << r2.current(v) * 1000 << " mA, P = " << r2.power(v) << " W";
    cout << (r2.power(v) > r2.watt ? "  ⚠️ 超過額定功率！" : "  ✅") << endl;
    return 0;
}
````

### 幻燈片 39：成員函式定義在類別外

# ✂️ 範圍解析運算子 `::` (Scope Resolution Operator)

類別內只寫**宣告**，實作寫在類別外，格式：`回傳型態 類別名稱::函式名稱(參數) { ... }`

大型專案通常把宣告放在 `.h` 標頭檔、實作放在 `.cpp` 檔，就像 IC 的「腳位定義」和「內部電路」分開。

````python
%%cuda
#include <iostream>
using namespace std;

class LED {
  public:
    string color;
    double forwardV;      // 順向電壓 Vf (V)
    double currentA;      // 工作電流 (A)

    double seriesResistor(double vcc);   // 只有宣告
};

// 類別外的實作：限流電阻 R = (Vcc - Vf) / I
double LED::seriesResistor(double vcc) {
    return (vcc - forwardV) / currentA;
}

int main() {
    LED red;
    red.color = "紅色"; red.forwardV = 2.0; red.currentA = 0.02;

    LED blue;
    blue.color = "藍色"; blue.forwardV = 3.2; blue.currentA = 0.02;

    cout << red.color  << " LED 接 5V 需要限流電阻 " << red.seriesResistor(5.0)  << " Ω" << endl;
    cout << blue.color << " LED 接 5V 需要限流電阻 " << blue.seriesResistor(5.0) << " Ω" << endl;
    return 0;
}
````

### 幻燈片 40：建構子

# 🏭 建構子 (Constructor)

- 物件**建立時自動執行**的特殊函式
- 名稱與類別**相同**，**沒有回傳型態**（連 void 都不寫）
- 可以有參數、也可以多載 (Overloading)
- 推薦用**成員初始化列表 (Member Initializer List)**：`: name(n), voltage(v)`

#### 💡 電機系比喻
就像工廠出廠設定 (Factory Default)：電池一出廠就決定好標稱電壓與容量。

````python
%%cuda
#include <iostream>
using namespace std;

class Battery {
  public:
    string model;
    double voltage;      // 標稱電壓 (V)
    int    capacity;     // 容量 (mAh)

    // 預設建構子 (Default Constructor)
    Battery() : model("未知型號"), voltage(0), capacity(0) {
        cout << "[建構] 建立一顆空白電池" << endl;
    }

    // 帶參數的建構子 + 初始化列表
    Battery(string m, double v, int c) : model(m), voltage(v), capacity(c) {
        cout << "[建構] " << model << " 出廠" << endl;
    }

    // 估算續航時間 (小時) = 容量 / 負載電流
    double runtimeHours(int loadmA) {
        return (double)capacity / loadmA;
    }

    // 能量 (Wh) = V × Ah
    double energyWh() {
        return voltage * capacity / 1000.0;
    }
};

int main() {
    Battery unknown;
    Battery cell("18650", 3.7, 3000);
    Battery aa("AA 鹼性電池", 1.5, 2500);

    cout << endl;
    cout << cell.model << "：能量 " << cell.energyWh() << " Wh，"
         << "負載 500mA 可用 " << cell.runtimeHours(500) << " 小時" << endl;
    cout << aa.model << "：能量 " << aa.energyWh() << " Wh，"
         << "負載 100mA 可用 " << aa.runtimeHours(100) << " 小時" << endl;
    return 0;
}
````

### 幻燈片 41：存取修飾子與封裝

# 🔒 存取修飾子 (Access Specifiers) 與封裝 (Encapsulation)

| 修飾子 | 類別內部 | 衍生類別 (子類別) | 外部 (main 等) |
|------|------|------|------|
| `public` | ✅ | ✅ | ✅ |
| `protected` | ✅ | ✅ | ❌ |
| `private`（class 預設） | ✅ | ❌ | ❌ |

**封裝**：把敏感資料設為 `private`，只能透過 `public` 的 **getter / setter** 存取，並在 setter 裡做檢查。

#### ⚡ 電機系範例：馬達控制器 (Motor Controller)
轉速 rpm 設為 private，`setSpeed()` 會自動限制在 0 ~ 最大轉速，避免有人直接設定超速把馬達燒掉。

````python
%%cuda
#include <iostream>
using namespace std;

class Motor {
  private:
    int rpm;             // 目前轉速：外部不能直接改
    int maxRpm;          // 最大轉速

  public:
    Motor(int maxR) : rpm(0), maxRpm(maxR) {}

    // Setter：寫入前先檢查（保護電路的概念）
    void setSpeed(int r) {
        if (r < 0) {
            cout << "  [拒絕] 轉速不可為負，設為 0" << endl;
            r = 0;
        } else if (r > maxRpm) {
            cout << "  [限制] " << r << " 超過上限，限制為 " << maxRpm << endl;
            r = maxRpm;
        }
        rpm = r;
    }

    // Getter：只讀
    int getSpeed() { return rpm; }
};

int main() {
    Motor fan(3000);

    fan.setSpeed(1500);
    cout << "目前轉速：" << fan.getSpeed() << " rpm" << endl;

    fan.setSpeed(5000);
    cout << "目前轉速：" << fan.getSpeed() << " rpm" << endl;

    fan.setSpeed(-100);
    cout << "目前轉速：" << fan.getSpeed() << " rpm" << endl;

    // fan.rpm = 99999;   // ❌ 編譯錯誤：rpm 是 private
    return 0;
}
````

### 幻燈片 42：階段五測驗

# 🔍 找錯誤：碰不到的成員

#### 這段程式為什麼編譯失敗？要怎麼改才符合「封裝」精神？

```cpp
class PowerSupply {
    double voltage;
  public:
    PowerSupply(double v) : voltage(v) {}
};

int main() {
    PowerSupply psu(12.0);
    psu.voltage = 48.0;
    cout << psu.voltage << endl;
}
```

````python
%%cuda
#include <iostream>
using namespace std;

class PowerSupply {
    double voltage;
  public:
    PowerSupply(double v) : voltage(v) {}
};

int main() {
    PowerSupply psu(12.0);
    psu.voltage = 48.0;
    cout << psu.voltage << endl;
    return 0;
}
````

---

#### ✅ 解答

**問題：** `class` 的成員**預設是 `private`**！`voltage` 寫在 `public:` 上方，所以外部無法存取 → `'double PowerSupply::voltage' is private within this context`。

**符合封裝的修正：** 不要把 voltage 改成 public，而是提供有檢查的 setter / getter：

```cpp
class PowerSupply {
    double voltage;
  public:
    PowerSupply(double v) : voltage(v) {}
    void setVoltage(double v) { if (v >= 0 && v <= 30) voltage = v; }  // 這台電源最高 30V
    double getVoltage() { return voltage; }
};
```

#### 💡 就像電源供應器的面板有「限壓旋鈕」：使用者只能在安全範圍內調整，而不是直接把變壓器拆開來改。

## 第六階段：繼承、多型與檔案處理

### 幻燈片 43：第六章封面

# 🧬 繼承 (Inheritance)、多型 (Polymorphism) 與檔案 (File I/O)

| 主題 | 說明 | 電機情境 |
|------|------|------|
| 單一繼承 | 子類別繼承父類別 | 元件 → 電阻 / 電容 |
| 多層繼承 | 祖父 → 父 → 子 | 裝置 → 感測器 → 溫度感測器 |
| 多重繼承 | 同時繼承多個父類別 | 智慧電表 = 電表 + WiFi 模組 |
| protected | 子類別可存取 | 共用校正參數 |
| 多型 virtual | 同一介面、不同行為 | R / L / C 的阻抗 |
| fstream | 讀寫檔案 | 量測資料記錄器 (Data Logger) |

### 幻燈片 44：繼承

# 👨‍👧 繼承 (Inheritance)

- **父類別 / 基底類別 (Base Class)**：被繼承的類別
- **子類別 / 衍生類別 (Derived Class)**：繼承者，用 `class 子 : public 父`
- 子類別自動擁有父類別的 public / protected 成員，還能再加自己的

#### 💡 電機系比喻
所有電子元件 (Component) 都有「名稱」和「料號」，電阻**是一種 (is-a)** 元件，再多了「阻值」；電容也是一種元件，多了「容值」。

````python
%%cuda
#include <iostream>
using namespace std;

// 父類別 (Base Class)
class Component {
  public:
    string refDes;          // 元件代號 (Reference Designator)，如 R1、C3
    string package;         // 封裝 (Package)，如 0805、DIP

    void label() {
        cout << "[" << refDes << " / " << package << "] ";
    }
};

// 子類別 (Derived Class)
class Resistor : public Component {
  public:
    double ohm;
};

class Capacitor : public Component {
  public:
    double farad;
    double ratedV;          // 耐壓
};

int main() {
    Resistor r;
    r.refDes = "R1"; r.package = "0805"; r.ohm = 4700;     // refDes, package 繼承而來

    Capacitor c;
    c.refDes = "C3"; c.package = "Radial"; c.farad = 100e-6; c.ratedV = 25;

    r.label(); cout << "電阻 " << r.ohm << " Ω" << endl;
    c.label(); cout << "電容 " << c.farad * 1e6 << " μF / " << c.ratedV << " V" << endl;
    return 0;
}
````

### 幻燈片 45：多層繼承與多重繼承

# 🌳 多層繼承 (Multilevel) 與 🔀 多重繼承 (Multiple)

```
多層繼承：Device → Sensor → TempSensor        多重繼承：  Meter     WiFiModule
                                                            ╲        ╱
                                                            SmartMeter
```

````python
%%cuda
#include <iostream>
using namespace std;

// ===== 多層繼承 (Multilevel Inheritance) =====
class Device {
  public:
    void powerOn() { cout << "裝置通電 (Power On)" << endl; }
};

class Sensor : public Device {
  public:
    void sample() { cout << "感測器取樣 (Sampling)" << endl; }
};

class TempSensor : public Sensor {
  public:
    void showTemp() { cout << "溫度 = 25.6 °C" << endl; }
};

// ===== 多重繼承 (Multiple Inheritance) =====
class Meter {
  public:
    double readKwh() { return 1234.5; }
};

class WiFiModule {
  public:
    void upload(double value) { cout << "透過 WiFi 上傳資料：" << value << " kWh" << endl; }
};

class SmartMeter : public Meter, public WiFiModule {
  public:
    void report() { upload(readKwh()); }   // 同時使用兩個父類別的功能
};

int main() {
    cout << "--- 多層繼承 ---" << endl;
    TempSensor ts;
    ts.powerOn();       // 來自祖父類別 Device
    ts.sample();        // 來自父類別 Sensor
    ts.showTemp();      // 自己的

    cout << "\n--- 多重繼承 ---" << endl;
    SmartMeter sm;
    sm.report();
    return 0;
}
````

### 幻燈片 46：protected 存取修飾子

# 🛡️ protected：自家人才能用

`protected` 成員：外部不能碰，但**子類別可以存取**。

#### ⚡ 電機系範例：感測器的校正係數
所有感測器共用「增益 (Gain)」與「偏移 (Offset)」校正參數，子類別在轉換數值時要用到，但不該讓外部隨意修改。

````python
%%cuda
#include <iostream>
using namespace std;

class AnalogSensor {
  protected:                 // 子類別可用，外部不行
    double gain   = 1.0;
    double offset = 0.0;
  public:
    void calibrate(double g, double o) { gain = g; offset = o; }
};

// LM35 溫度感測器：10 mV / °C
class LM35 : public AnalogSensor {
  public:
    double toCelsius(double mv) {
        return (mv / 10.0) * gain + offset;   // 直接使用 protected 成員
    }
};

int main() {
    LM35 sensor;
    cout << "未校正：250 mV → " << sensor.toCelsius(250) << " °C" << endl;

    sensor.calibrate(1.04, -0.3);
    cout << "校正後：250 mV → " << sensor.toCelsius(250) << " °C" << endl;

    // sensor.gain = 5;   // ❌ 編譯錯誤：gain 是 protected
    return 0;
}
````

### 幻燈片 47：多型 Polymorphism

# 🎭 多型：同一個介面，不同的行為

**多型 (Polymorphism)** = "many forms"。用**父類別指標**指向不同的子類別物件，呼叫同一個函式卻得到各自的結果。

關鍵字：
- `virtual`：父類別的函式加上它，才會依照「實際物件」決定呼叫哪個版本（動態繫結 Dynamic Binding）
- `override`：子類別標註「我要覆寫父類別的 virtual 函式」，打錯字時編譯器會提醒
- `= 0`：純虛擬函式 (Pure Virtual Function)，該類別變成抽象類別 (Abstract Class)，不能直接建立物件

#### ⚡ 電機系範例：R、L、C 在交流電路中的阻抗 (Impedance)

| 元件 | 阻抗大小 \|Z\| |
|------|------|
| 電阻 R | $R$（與頻率無關） |
| 電感 L | $X_L = 2\pi f L$（頻率越高越大） |
| 電容 C | $X_C = \dfrac{1}{2\pi f C}$（頻率越高越小） |

#### 先看「沒有 virtual」會發生什麼事

父類別指標只會呼叫**父類別版本**，因為編譯器在編譯時就決定好了（靜態繫結 Static Binding）。

````python
%%cuda
#include <iostream>
using namespace std;

class Part {
  public:
    string describe() { return "一般元件"; }          // 沒有 virtual
};
class Inductor : public Part {
  public:
    string describe() { return "電感 (Inductor)"; }
};

int main() {
    Inductor L1;
    Part *p = &L1;                                     // 父類別指標指向子類別物件
    cout << "直接呼叫：" << L1.describe() << endl;      // 電感
    cout << "透過父類別指標：" << p->describe() << endl; // 一般元件 ← 不是我們要的！
    return 0;
}
````

#### 加上 virtual：真正的多型
`p->impedance(f)` 中的 `->` 等同於 `(*p).impedance(f)`，用指標存取物件成員時使用。

````python
%%cuda
#include <iostream>
#include <iomanip>
#include <vector>
using namespace std;

const double PI = 3.14159265358979;

// 抽象類別 (Abstract Class)
class Element {
  public:
    string name;
    Element(string n) : name(n) {}
    virtual double impedance(double f) = 0;   // 純虛擬函式
    virtual ~Element() {}                     // 虛擬解構子 (Virtual Destructor)
};

class Res : public Element {
    double R;
  public:
    Res(string n, double r) : Element(n), R(r) {}
    double impedance(double f) override { return R; }
};

class Ind : public Element {
    double L;
  public:
    Ind(string n, double l) : Element(n), L(l) {}
    double impedance(double f) override { return 2 * PI * f * L; }
};

class Cap : public Element {
    double C;
  public:
    Cap(string n, double c) : Element(n), C(c) {}
    double impedance(double f) override { return 1.0 / (2 * PI * f * C); }
};

int main() {
    // 同一個 vector 放不同種類的元件
    vector<Element*> parts = {
        new Res("R 100ohm", 100),
        new Ind("L 10mH", 10e-3),
        new Cap("C 10uF", 10e-6)
    };
    double freqs[] = {60, 1000, 10000};

    cout << setw(10) << "Part";
    for (double f : freqs) cout << setw(12) << (to_string((int)f) + "Hz");
    cout << endl << fixed << setprecision(2);

    for (Element *p : parts) {
        cout << setw(10) << p->name;
        for (double f : freqs) cout << setw(12) << p->impedance(f);   // 多型呼叫
        cout << endl;
    }

    for (Element *p : parts) delete p;      // 釋放記憶體
    return 0;
}
````

**觀察：** 電感的阻抗隨頻率**上升**、電容的阻抗隨頻率**下降** — 這就是低通 / 高通濾波器 (Low-pass / High-pass Filter) 的原理！

### 幻燈片 48：檔案讀寫 fstream

# 📂 `<fstream>` 檔案處理

| 類別 | 用途 | C 語言對應 |
|------|------|------|
| `ofstream` | 寫入檔案 (Output File Stream) | `fopen(..., "w")` + `fprintf` |
| `ifstream` | 讀取檔案 (Input File Stream) | `fopen(..., "r")` + `fscanf` |
| `ofstream f(name, ios::app)` | 附加 (Append) 模式 | `fopen(..., "a")` |

用法和 `cout` / `cin` 幾乎一樣，只是對象從螢幕 / 鍵盤換成檔案！

#### ⚡ 電機系範例：資料記錄器 (Data Logger)
把太陽能板每小時的電壓、電流存成 CSV 檔，再讀回來計算總發電量。CSV 檔可以直接用 Excel 開啟畫圖。

````python
%%cuda
#include <iostream>
#include <fstream>
#include <string>
using namespace std;

int main() {
    // ===== 1. 寫入 CSV =====
    ofstream fout("solar_log.csv");
    if (!fout) {                                  // 檢查是否開檔成功
        cout << "無法建立檔案！" << endl;
        return 1;
    }
    fout << "hour,voltage,current" << endl;       // 標題列
    double v[] = {17.2, 18.1, 18.6, 18.4, 17.5};
    double i[] = {2.1, 4.5, 5.6, 5.2, 3.0};
    for (int h = 0; h < 5; h++) {
        fout << h + 10 << "," << v[h] << "," << i[h] << endl;   // 10 點到 14 點
    }
    fout.close();
    cout << "已寫入 solar_log.csv" << endl;

    // ===== 2. 讀回並計算 =====
    ifstream fin("solar_log.csv");
    string header;
    getline(fin, header);                         // 跳過標題列

    int hour;
    double volt, amp, totalWh = 0;
    char comma;
    while (fin >> hour >> comma >> volt >> comma >> amp) {
        double watt = volt * amp;
        totalWh += watt * 1.0;                    // 每筆代表 1 小時
        cout << hour << ":00  " << volt << " V × " << amp << " A = " << watt << " W" << endl;
    }
    fin.close();

    cout << "總發電量 = " << totalWh << " Wh" << endl;
    return 0;
}
````

### 幻燈片 49：綜合專題 — 串聯 RLC 電路分析器

# 🛠️ 綜合專題：串聯 RLC 電路分析器

整合本課程學到的：**類別、建構子、封裝、vector、函式、迴圈、格式化輸出、檔案**。

串聯 RLC 電路：
- 總阻抗 $|Z| = \sqrt{R^2 + (X_L - X_C)^2}$
- 諧振頻率 (Resonant Frequency) $f_0 = \dfrac{1}{2\pi\sqrt{LC}}$，此時 $X_L = X_C$，電流最大
- 電流 $I = V / |Z|$

程式會掃描頻率 (Frequency Sweep)，畫出文字版的頻率響應圖，並把結果存成 CSV。

````python
%%cuda
#include <iostream>
#include <iomanip>
#include <fstream>
#include <vector>
#include <cmath>
using namespace std;

const double PI = 3.14159265358979;

class SeriesRLC {
  private:
    double R, L, C;

  public:
    SeriesRLC(double r, double l, double c) : R(r), L(l), C(c) {}

    double resonantFreq() const { return 1.0 / (2 * PI * sqrt(L * C)); }

    double impedance(double f) const {
        double XL = 2 * PI * f * L;
        double XC = 1.0 / (2 * PI * f * C);
        return sqrt(R * R + (XL - XC) * (XL - XC));
    }

    double current(double vrms, double f) const { return vrms / impedance(f); }
};

int main() {
    SeriesRLC circuit(10.0, 10e-3, 1e-6);     // R=10Ω, L=10mH, C=1μF
    double vrms = 1.0;
    double f0 = circuit.resonantFreq();

    cout << fixed << setprecision(1);
    cout << "諧振頻率 f0 = " << f0 << " Hz" << endl << endl;

    // 產生掃頻點：f0 的 0.25 倍 ~ 4 倍
    vector<double> freqs;
    for (double k = 0.25; k <= 4.01; k *= sqrt(2.0)) freqs.push_back(f0 * k);

    ofstream csv("rlc_sweep.csv");
    csv << "freq_Hz,Z_ohm,I_mA" << endl;

    cout << setw(10) << "f (Hz)" << setw(10) << "|Z| (Ω)" << setw(10) << "I (mA)" << "  電流大小" << endl;
    for (double f : freqs) {
        double z = circuit.impedance(f);
        double i = circuit.current(vrms, f) * 1000;
        csv << f << "," << z << "," << i << endl;

        cout << setw(10) << f << setw(10) << z << setw(10) << i << "  ";
        for (int n = 0; n < (int)(i / 4); n++) cout << "▇";
        cout << endl;
    }
    csv.close();
    cout << "\n掃頻結果已存成 rlc_sweep.csv" << endl;
    return 0;
}
````

#### 📊 加分題：用 Python 把 CSV 畫成圖
Colab 同時也是 Python 環境，可以直接讀取剛才 C++ 產生的 CSV 檔畫出頻率響應曲線。

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("rlc_sweep.csv")
plt.semilogx(df["freq_Hz"], df["I_mA"], marker="o")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Current (mA)")
plt.title("Series RLC Frequency Response")
plt.grid(True, which="both")
plt.show()
````
